```python
from bs4 import BeautifulSoup

with open("scrape_data/data1.html","r") as f:
    html_content = f.read() 

soup = BeautifulSoup(html_content, "html.parser")

print(soup.head.link['href'])
```

```python
all_h3 = soup.find_all('h3')
countries = []
for h3 in all_h3:
    name = h3.get_text(strip=True)
    population = h3.find_next("div").select("span.country-population")[0].get_text(strip=True)
    # print(h3.find_parent("div").find_all("div"))
    countries.append([name,population])

```

```python
import pandas as pd

df = pd.DataFrame(countries,columns=["Name","Population"])
df.to_csv("scrape_data/clean_data.csv",index=False)
print(df)
```
