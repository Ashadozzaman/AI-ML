```python
movies = [
    "Avatar",
    "Titanic",
    "Avengers",
    "Jurassic Park",
    "The Lion King",
    "Frozen",
    "Barbie",
    "Top Gun",
    "Iron Man",
    "Harry Potter"
]

oscar_movies = [
    "Gladiator",
    "A Beautiful Mind",
    "Chicago",
    "The Lord of the Rings: The Return of the King",
    "Million Dollar Baby",
    "Crash",
    "The Departed",
    "No Country for Old Men",
    "Slumdog Millionaire",
    "The Hurt Locker",
]

oscar_years = list(range(2000, 2024))

years = [2009, 1997, 2019, 1993, 2019, 2013, 2023, 2022, 2008, 2001]

earnings = [2923, 2264, 2799, 1046, 1657, 1453, 1446, 1495, 585, 974]
oscar_earnings = [2923, 2264, 2799, 1046, 1657, 1453, 1446, 1495, 585, 974]

# plot here
plt.plot(years, earnings,marker = 'o')
plt.title("Movie Earning per years")
plt.xlabel("Year")
plt.ylabel("Earning")
plt.grid(True)
plt.show()
```

```python
plt.plot(years, earnings, marker='o')

for movie, year, earning in zip(movies, years, earnings):
    plt.annotate(
        movie,
        (year, earning),
        textcoords="offset points",
        xytext=(0, 10),
        ha='center'
    )

plt.xlabel("Year")
plt.ylabel("Earnings (Million USD)")
plt.title("Movie Earnings")
plt.grid(True)
plt.show()
```

```python
movies = [
    "Avatar",
    "Titanic",
    "Avengers",
    "Jurassic Park",
    "The Lion King",
    "Frozen",
    "Barbie",
    "Top Gun",
    "Iron Man",
    "Harry Potter"
]

oscar_movies = [
    "Gladiator",
    "A Beautiful Mind",
    "Chicago",
    "The Lord of the Rings: The Return of the King",
    "Million Dollar Baby",
    "Crash",
    "The Departed",
    "No Country for Old Men",
    "Slumdog Millionaire",
    "The Hurt Locker",
]

years = range(2010,2020)

earnings = [2923, 2264, 2799, 1046, 1657, 1453, 1446, 1495, 585, 974]

oscar_earnings = [1800, 1200, 1500, 900, 1300, 1100, 800, 950, 500, 700]

plt.plot(years, earnings, "o--", label='Non-Oscar Movies')
plt.plot(years, oscar_earnings, marker='s', label='Oscar Movies')

plt.title("Non-Oscar vs Oscar Movie Earnings")
plt.xlabel("Year")
plt.ylabel("Earnings ($ Million)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.bar(years, earnings)

plt.title("Movie Earnings")
plt.xlabel("Year")
plt.ylabel("Earnings ($ Million)")
for i in range(len(years)):
    plt.text(years[i],earnings[i]+15,str(earnings[i]), ha="center")

plt.show() 
```
