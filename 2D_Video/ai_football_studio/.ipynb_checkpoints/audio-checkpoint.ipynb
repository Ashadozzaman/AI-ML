```python
import json

with open("ai_football_discussion.json", "r", encoding="utf-8") as file:
    discussion = json.load(file)

conversation = discussion["conversation"]
```

```python
VOICE_MAP = {
    "Rahim": "bn-BD-PradeepNeural",
    "Karim": "bn-IN-BashkarNeural",
    "Jamal": "bn-BD-PradeepNeural"
}
```

```python
import edge_tts
from pathlib import Path

AUDIO_DIR = Path("audio")
AUDIO_DIR.mkdir(exist_ok=True)

async def generate_audio(text, voice, output_file):

    communicate = edge_tts.Communicate(
        text=text,
        voice=voice,
        rate="+0%",
        pitch="+0Hz"
    )

    await communicate.save(str(output_file))

    print(f"Audio generated: {output_file}")
```

```python
from PIL import Image, ImageDraw
from pathlib import Path

PRESENTER_DIR = Path("presenters")
VIDEO_DIR = Path("video")

PRESENTER_DIR.mkdir(exist_ok=True)
VIDEO_DIR.mkdir(exist_ok=True)
```

```python
import subprocess

for index, line in enumerate(conversation):

    speaker = line["speaker"]

    image_file = PRESENTER_DIR / f"{speaker}.png"
    audio_file = AUDIO_DIR / f"{index:03d}_{speaker}.mp3"

    output_file = VIDEO_DIR / f"{index:03d}.mp4"

    command = [
        "ffmpeg",
        "-y",
        "-loop", "1",
        "-framerate", "25",
        "-i", str(image_file),
        "-i", str(audio_file),
        "-vf",
        "scale=1280:720:force_original_aspect_ratio=increase,crop=1280:720,format=yuv420p",
        "-c:v", "libx264",
        "-preset", "ultrafast",
        "-tune", "stillimage",
        "-c:a", "aac",
        "-b:a", "192k",
        "-shortest",
        "-r", "25",
        str(output_file)
    ]

    subprocess.run(command, check=True)

    print("Created:", output_file)
```
