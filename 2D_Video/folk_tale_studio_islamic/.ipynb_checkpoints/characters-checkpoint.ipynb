# characters · see and change how people look
Pick a story, see its characters, then change their look in the story's `CHARACTERS:` lines (colours, hats, beards...). Re-run to preview.

```python
STORY = "stories/5_counting_the_crows.txt"
from engine.script import load_story
from engine.rig import Rig
from engine import art
from PIL import Image
from IPython.display import display

story = load_story(STORY)
bg = art.make_rajshobha(False)
rigs = {n: Rig(n, art.design(c["design"])) for n, c in story["characters"].items()}
n = len(rigs)
frame = bg.copy()
for i, (name, rig) in enumerate(rigs.items()):
    rig.draw(frame, int(1280 * (i + 1) / (n + 1)), 596, dict(mouth="smile", brow="happy", arm_r=(30, 20)))
display(frame)
```

### Look settings (put them after the voice settings in a `CHARACTERS:` line)

| key | example | meaning |
|---|---|---|
| `skin`, `hair`, `coat`, `trim`, `pants`, `shoes` | `coat=20,100,150` | colours as red,green,blue (0-255) |
| `hat` | `hat=turban` | `crown`, `turban`, `jester`, `topi` (white cap), `none` (bare head), `tiki` (bald + tuft), `pagri` (royal), `pagri_tall`, `bun` (Ginni's hair) |
| `hat_main`, `hat_accent` | `hat_main=214,110,40` | hat colours |
| `beard` | `beard=none` | `none`, `mustache`, `goatee`, `handlebar` |
| `face_w`, `face_h`, `nose`, `build`, `scale` | `build=1.2 scale=0.68` | face width/height, nose size, body width, overall size |
| `style` | `style=tunic` | `kurta` (long coat), `tunic` (short, zig-zag) or `dhoti` (bare chest + shoulder cloth; `coat` = the cloth) or `saree` (`coat` = saree, `trim` = border, `blouse` = blouse colour) |

Base looks: `chikna`, `ginni`, `krishnachandra`, `mantri` (the series cast) plus `king`, `minister`, `jester`. Extra switches: `tilak=true` (forehead mark), `bindi`, `sindoor`, `earrings`, `nath` (for Ginni).
