## Make your own story
Copy `story.json`, change the text and beats, and render again. Beat types:

| beat | example |
|---|---|
| `title` / `end` | `{"do":"title","text":"...","sub":"...","dur":3}` |
| `scene` | `{"do":"scene","bg":"garden","cast":{"raja":{"x":0.4,"face":"right"}}}` |
| `enter` / `exit` / `move` | `{"do":"enter","who":"chandu","from":"right","to":0.6,"dur":2}` |
| `say` | `{"do":"say","who":"raja","text":"...","emotion":"angry","gesture":"point","to":"chandu","also":{"mantri":"nervous"}}` |
| `act` | `{"do":"act","who":"chandu","action":"bow","dur":1.5}` |
| `wait`, `set`, `dark`, `prop`, `cam` | see `engine/timeline.py` |

* **Backgrounds:** `garden`, `garden_empty`, `court`, `court_evening`
* **Emotions:** `neutral happy confident angry scared nervous surprised sad thinking scheming smirk`
* **Gestures (while talking):** `open chest raise point shrug fist palm wave show_hands salute`
* **Actions:** `salute bow shock laugh think cheer nervous point show_hands guilty stomp smirk wave touch cower`
* `x` is the position across the screen: 0 = left edge, 1 = right edge.
* Background music: put an `music/music.mp3` file in the project - it is mixed in quietly.
