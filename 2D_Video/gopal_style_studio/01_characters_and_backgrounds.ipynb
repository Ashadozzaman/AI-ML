## 1 · The three characters (separate parts = cut-out puppets)

````python
from IPython.display import display
from PIL import Image
from engine import art
from engine.rig import Rig

bg = Image.open('assets/bg_court.png').convert('RGB') if __import__('os').path.exists('assets/bg_court.png') else art.make_court()
frame = bg.copy()
poses = {
    'raja':   (260,  dict(mouth='open', brow='angry', arm_r=(70, 20), hand_r='point')),
    'mantri': (640,  dict(mouth='smile', torso=3, head=-5, arm_l=(30, 70), hand_l='fist')),
    'chandu': (1010, dict(mouth='wide', eyes='happy', brow='happy', arm_l=(120, 30), arm_r=(120, 30), leg_l=14, leg_r=-14)),
}
designs = {'raja': 'king', 'mantri': 'minister', 'chandu': 'jester'}
for name, (x, pose) in poses.items():
    Rig(name, art.design(designs[name])).draw(frame, x, 596, pose)
display(frame)
````

### Change how a character looks
Open `story.json` and edit the `"design"` of a character, for example:
```json
"raja": {"design": {"base": "king", "skin": [200,140,100], "coat": [20,60,140], "hat": "turban", "beard": "none"}}
```
Useful keys: `skin`, `hair`, `coat`, `trim`, `pants`, `shoes`, `hat` (`crown` / `turban` / `jester`), `hat_main`, `hat_accent`, `beard` (`none` / `mustache` / `goatee`), `face_w`, `face_h`, `nose`, `build`, `scale`.  
Always use **original** characters - do not copy the look of the TV show.
