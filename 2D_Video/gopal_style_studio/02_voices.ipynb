# 02 · Voices
Creates one audio file per spoken line in `audio/` (`line_01.mp3`, `line_02.mp3`, ...).

* **Real voices:** `edge-tts` (free Microsoft voices, needs internet). Install once: `pip install edge-tts`
* **Draft voices:** an offline humming sound, only to test timing and lip sync.

Each character has its own voice in `story.json` (`name`, `rate`, `pitch`). Emotions (`angry`, `scared`, `happy`...) change speed and pitch a little.

```python
import json
from engine import audio

story = json.load(open('story.json', encoding='utf-8'))
for lid, beat in audio.say_beats(story):
    print(lid, beat['who'], '-', beat['text'][:50])
```

## Bangla voices available in edge-tts (run once to see the list)

```python
try:
    import edge_tts
    voices = await edge_tts.list_voices()
    for v in voices:
        if v['Locale'].startswith('bn-'):
            print(v['ShortName'], v['Gender'])
except Exception as e:
    print('Could not load the voice list:', e)
```

## Make the voices
Set `overwrite=True` if you changed the text or voice settings and want to remake every file.

```python
try:
    await audio.tts_all(story, 'audio', overwrite=False)
except Exception as e:
    print('\nReal voices are not available right now:', e)
    print('Making offline DRAFT voices so you can still test the animation...')
    audio.make_all_draft_audio(story, 'audio')
```

## Listen

```python
from IPython.display import Audio, display
from pathlib import Path
for lid, beat in audio.say_beats(story):
    p = audio.find_audio('audio', lid)
    print(lid, beat['who'], p.name if p else 'MISSING')
    if p and lid in ('line_01', 'line_02'):
        display(Audio(str(p)))
```
